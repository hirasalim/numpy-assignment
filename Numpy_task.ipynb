task 5 pssing marks are 50 find

```python
import numpy as np

marks = np.array([35, 67, 89, 45, 90, 22, 70, 58])

passed = marks[marks >= 50]
failed = marks[marks < 50]

print("Passed:", passed)
print("Failed:", failed)
print("Passed students:", len(passed))
print("Failed students:", len(failed))
```

```python

#IMPORT NUMPY

import numpy as np
# PYTHON list
marks = [60,20,50,90,30]
print(marks)
#NUMPY LIST
marks = np.array([70,30,40,90,80])
print(marks + 5)
```

```python
# ONE DIMENSIONAL AARAY
numbers = np.array([10, 20, 30, 40])

print(numbers)
#two dimension aray
marks = np.array([
    [10,20,30,40,50],
    [50,60,70,80,90]
    ])
print(marks)

```

```python
#array properties in numpy
marks = np.array([
    [70,30,40,90,80],
    [40,50,30,70,90],
    [90,40,80,70,50]
    ])
print(marks.ndim)
print(marks.shape)
print(marks.size)
print(marks.dtype)
```

```python

#indexing
marks = np.array([70,30,40,90,80])
print(marks[0])

#NEGATIVE INDEXING
print(marks[-2])


# indexing in 2d

marks = np.array([
    [20,60,50,70,80],
    [70,50,10,30,90]
    ])
print(marks[0, 2])
print(marks[1, -2])

# slicing
marks = np.array([80,30,60,90,70])
print(marks[1:4])

marks = np.array([
    [70,30,40,90,80],
    [40,50,30,70,90]
    ])
print(marks[1,1:4])

# VACTORIZED OPERATION
numbers = np.array([10, 20, 30, 40])
print(numbers + 5)
print(numbers - 5)
print(numbers * 2)
print(numbers / 2)




```

```python
# OPERATION BETWEEN TWO ARRAY
a = np.array([10, 20, 30])

b = np.array([1, 2, 3])
print(a + b)
```

```python
# MATHEMATICAL FUNCTION

numbers = np.array([4, 9, 16, 25])

print(np.sqrt(numbers))
print(np.power(numbers, 2))
print(np.abs(numbers))
```

```python
#round number
numbers = np.array([4.2, 5.7, 8.9])

print(np.round(numbers))
```

```python
#aggergation function
marks = np.array([70, 85, 60, 95, 80])
print(np.sum(marks))
print(np.min(marks))
print(np.max(marks))
print(np.mean(marks))
print(np.median(marks))
print(np.std(marks))
print(np.var(marks))
```

```python
# CONDITION in numpy
marks = np.array([30, 50, 80, 45, 90, 20])

print(marks >= 50)

print(marks >= 50)
```

```python
# filtering data
marks = np.array([30, 50, 80, 45, 90, 20])

passed = marks[marks >= 50]

print(passed)
# failed
failed = marks[marks < 50]

print(failed)
```

```python
# reshape
numbers = np.array([1, 2, 3, 4, 5, 6])
new_array = numbers.reshape(2, 3)

print(new_array)
```

```python
# FLATTEN DATA
## `flatten changes a multi-dimensional array into a one-dimensional array.

data = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

print(data.flatten())
```

```python
# USEFULL NUMPY OPERATION
print(np.zeros(5))
print(np.ones(5))
print(np.arange(1, 11))
marks = np.random.randint(40, 100, 10)

print(marks)
```
